# Display Fibonacci series up to 10 terms

```python
# Create two variables for storing values, set variables values as 0 and 1 in the start
num1 = 0
num2 = 1

print("The Fibonacci sequence is")

# Create a for loop to print first 10 fibonacci numbers

for i in range(10):

    #Print the num1 value with a space
    print(num1, end=' ')
    result = num1 + num2

    #Swapping the values
    num1 = num2
    num2 = result
```

```python
# In fibonnaci series the sequence is calculated using a simple recursive rule where each number is the sum of the two preceding ones. The sequence typically starts with 0 and 1.

# Hence num1 is set to 0 and num2 is set to 1. So, when after first iteration, num1 will have value of num2 that is 1 and num2 will have value of result that is also 1 and num1 will print value 1.

# In second iteration num 1 will have value of 1 and num2 will have value 1, so, now result is 2.

#Fibonacci sequence calculation:
 #num1 = 0
 # 0 + 1 =  1
 # 1 + 1 =  2
 # 1 + 2 =  3
 # 2 + 3 =  5
 # 3 + 5 =  8
 # 5 + 8 =  13
 # 8 + 13 = 21
 # 13 + 21 = 34
```

# Find the sum of a series of a number up to n terms

```python
num = 2
terms = 5
count = 0
list1 = []

while terms > 0:
    count = (count * 10) + num
    list1.append(count)
    terms = terms - 1

result = sum(list1[:])
print(result)
```

# Reverse a integer number

```python
num = 76542
reverse_num = 0
print("Given Number", num)
while num > 0:
    reminder = num % 10
    reverse_num = (reverse_num * 10) + reminder
    num = num // 10

print("Reverse Number", reverse_num)
```

# Print the following pattern

```python
row = 5

for i in range(0,row):
    for j in range(0, i+1):
        print("*", end= ' ')
    print("\r")

for i in range(row,0,-1):
    for j in range(0, i - 1):
        print("*", end = ' ')
    print("\r")
```

# Find largest and smallest digit in a number

```python
num1 = int(input("Please enter a number"))
if num1 == 0:
    print("The number is 0 and the largest and the smallest number is 0")

s_num = str(abs(num1))
largest_num = int(s_num[0])
smallest_num = int(s_num[0])

for i in s_num[1:]:
    digi_int = int(i)
    if digi_int > largest_num:
        largest_num = digi_int

    if digi_int < smallest_num:
        smallest_num = digi_int

print("Largest number: {}".format(largest_num))
print("Smallest number: {}".format(smallest_num))
```
